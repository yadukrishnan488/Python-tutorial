```python
# 1. Sum and Difference of Two Numbers

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
print("Sum =", a + b)
print("Difference =", a - b)
```

```python

# 2. Product Price and Quantity
price = float(input("Enter product price: "))
quantity = int(input("Enter quantity: "))
total = price * quantity
print("Total cost = ₹", total)
```

```python
# 3. Average Marks
marks = float(input("Enter total marks: "))
subjects = int(input("Enter number of subjects: "))
average = marks / subjects
print("Average marks =", average)
```

```python
# 4. Remainder When Divided by 5
number = int(input("Enter an integer: "))
remainder = number % 5
print("Remainder =", remainder)
```

```python
# 5. Complete Groups and Students Left Over
students = 47
group_size = 6
groups = students // group_size
leftover = students % group_size
print("Complete groups =", groups)
print("Students left over =", leftover)
```

```python
# 6. Area of a Square Using **
side = float(input("Enter side of square: "))
area = side ** 2
print("Area of square =", area)
```

```python
# 7. Total Customer Bill
notebooks = 3 * 45
pens = 2 * 20
bag = 1 * 750
total = notebooks + pens + bag
print("Total bill = ₹", total)
```

```python
# 8. Convert Seconds into Hours, Minutes and Seconds
seconds = int(input("Enter seconds: "))
hours = seconds // 3600
minutes = (seconds % 3600) // 60
remaining_seconds = seconds % 60
print("Hours =", hours)
print("Minutes =", minutes)
print("Seconds =", remaining_seconds)
```

```python
# 9. Extract Hundreds, Tens and Units Digits
number = int(input("Enter a three-digit number: "))
hundreds = number // 100
tens = (number // 10) % 10
units = number % 10
print("Hundreds digit =", hundreds)
print("Tens digit =", tens)
print("Units digit =", units)
```

```python

# 10. Calculate Final Salary
basic_salary = float(input("Enter basic salary: ₹"))
hra = float(input("Enter HRA: ₹"))
travel = float(input("Enter travel allowance: ₹"))
deduction = float(input("Enter deduction: ₹"))
final_salary = basic_salary + hra + travel - deduction
print("Final salary = ₹", final_salary)
```
