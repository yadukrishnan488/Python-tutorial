# Mini Project 1 – Purchase Bill

Five separate test inputs. Run each code cell and take screenshots.

## Test 1 – Invalid Input

```python
item_name = "Pen"
price = -20
quantity = 2
member = "no"

if price <= 0 or quantity <= 0:
    print("Error: Price and quantity must be greater than 0.")
    print("Bill cannot be generated.")
else:
    subtotal = price * quantity

    if subtotal >= 5000:
        discount_rate = 10
    elif subtotal >= 2000:
        discount_rate = 5
    else:
        discount_rate = 0

    if member == "yes":
        discount_rate = discount_rate + 2

    discount = subtotal * discount_rate / 100
    amount_after_discount = subtotal - discount
    gst = amount_after_discount * 18 / 100
    amount_before_delivery = amount_after_discount + gst

    if amount_before_delivery > 1000:
        delivery = 0
        delivery_text = "Free"
    else:
        delivery = 50
        delivery_text = "Rs. 50.00"

    total = amount_before_delivery + delivery

    print("------ BILL ------")
    print(f"Item       : {item_name}")
    print(f"Subtotal   : {subtotal:.2f}")
    print(f"Discount   : {discount:.2f} ({discount_rate}%)")
    print(f"GST (18%)  : {gst:.2f}")
    print(f"Delivery   : {delivery_text}")
    print(f"TOTAL      : {total:.2f}")
    print("------------------")
```
