**22. Employee Salary Calculation

Input an employee's basic salary.

Calculate allowances based on salary:

●	Salary < ₹20,000 → HRA = 10%, DA = 5%

●	₹20,000–₹49,999 → HRA = 15%, DA = 10%

●	₹50,000–₹99,999 → HRA = 20%, DA = 15%

●	₹1,00,000 or above → HRA = 25%, DA = 20%

Calculate:

Gross Salary = Basic Salary + HRA + DA

Display the salary details.
**

```python
# Employee Salary Calculation

# Input basic salary
basic_salary = float(input("Enter basic salary: ₹"))

# Calculate HRA and DA based on salary
if basic_salary < 20000:
    hra = basic_salary * 10 / 100
    da = basic_salary * 5 / 100

elif basic_salary < 50000:
    hra = basic_salary * 15 / 100
    da = basic_salary * 10 / 100

elif basic_salary < 100000:
    hra = basic_salary * 20 / 100
    da = basic_salary * 15 / 100

else:
    hra = basic_salary * 25 / 100
    da = basic_salary * 20 / 100

# Calculate gross salary
gross_salary = basic_salary + hra + da

# Display salary details
print("\n--- Salary Details ---")
print("Basic Salary: ₹", basic_salary)
print("HRA: ₹", hra)
print("DA: ₹", da)
print("Gross Salary: ₹", gross_salary)
```
